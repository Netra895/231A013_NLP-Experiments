```python
print("Name: Netra Neman", "UIN: 231A013")
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import UnigramTagger, BigramTagger, pos_tag
from nltk.corpus import treebank
from nltk import pos_tag_sents
```

```python
print("Name: Netra Neman", "UIN: 231A013")

nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')
nltk.download('treebank')
```

```python
print("Name: Netra Neman", "UIN: 231A013")
nltk.download('punkt_tab')
sample_text = "He is reading a book. Please book a ticket for me."
tokens = word_tokenize(sample_text)
```

```python
print("Name: Netra Neman", "UIN: 231A013")
nltk.download('averaged_perceptron_tagger_eng')
rule_based_tags = pos_tag(tokens)
print(rule_based_tags)
```

```python
print("Name: Netra Neman", "UIN: 231A013")
nltk.download('tagsets_json')
nltk.help.upenn_tagset('NNS')
```

```python
print("Name: Netra Neman", "UIN: 231A013")

train_sents = treebank.tagged_sents()
unigram_tagger = UnigramTagger(train_sents)
unigram_tag = unigram_tagger.tag(tokens)

bigram_tagger = BigramTagger(train_sents)
bigram_tag = bigram_tagger.tag(tokens)
```

```python
print("Name: Netra Neman", "UIN: 231A013")
print(unigram_tag)
```

```python
print("Name: Netra Neman", "UIN: 231A013")

print(bigram_tag)
```

```python
print("Name: Netra Neman", "UIN: 231A013")

print("Original sentence:", sample_text)
print("\nRule-based tagging(Statistical):")
print(rule_based_tags)
print("\nUnigram tagging:")
print(unigram_tag)
print("\nBigram tagging:")
print(bigram_tag)
```

```python
print("Name: Netra Neman", "UIN: 231A013")

test_sents = treebank.tagged_sents()[1000:]
unigram_acc = unigram_tagger.evaluate(test_sents)
bigram_acc = bigram_tagger.evaluate(test_sents)

print(f"\nUnigram accuracy: {unigram_acc* 100:.2f}%")
print(f"\nBigram accuracy: {bigram_acc* 100:.2f}%")
```
