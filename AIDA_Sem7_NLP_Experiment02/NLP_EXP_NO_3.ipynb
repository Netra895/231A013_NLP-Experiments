```python
print("Netra Neman" ,"UIN: 231A013\n")
text_lower = text.lower()
print(text_lower)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

import re
pattern = r'https?://www\.\S+|www\.\S+|\S\@\S+'
text_url = re.sub(pattern, " ", text_lower)
print(text_url)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

import re
pattern = r'\d+\S+'
text_num = re.sub(pattern, " ", text_url)
print(text_num)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")
import re
pattern = r'\d+\S+\#\S+'
text_num = re.sub(pattern, " ", text_num)
print(text_num)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

import emoji

text_emoji = emoji.replace_emoji(text_num, replace="")
print(text_emoji)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

import nltk
nltk.download("punkt_tab")
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

from nltk import word_tokenize
tokenized=word_tokenize(text_emoji)
print("Tokenized words: ",tokenized)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

print("With punctuation:", tokenized)
print()
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

import string
punctuation = list(string.punctuation)
print(punctuation)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

nltk.download("stopwords")
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

from nltk.corpus import stopwords
print((stopwords.words('english')))
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

# First, remove punctuation from the tokenized list
text_punct = [word for word in tokenized if word not in punctuation]

# Now, remove stopwords from the punctuation-cleaned list
text_clean2 = [word for word in text_punct if word not in stopwords.words("english")]
print("Cleaned text: ", text_clean2)
```

```python
print("Netra Neman" ,"UIN: 231A013\n")

from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in text_clean2]
print("Stemmed words: ", stemmed_words)
```
