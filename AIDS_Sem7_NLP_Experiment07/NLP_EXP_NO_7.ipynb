```python
print("Name: Netra Neman", "UIN: 231A013")

import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

nltk.download('punkt_tab')

text="""Netra Neman is a student from Mumbai, Maharashtra.She is pursuing Artificial Intelligence and Data Science at Rizvi College of Engineering.
She is interested in developing AI-based solutions and exploring new technologies.Her organisation focuses on innovation and technical learning."""

tokens=word_tokenize(text)
print(tokens)
```

```python
print("Name: Netra Neman", "UIN: 231A013")
nltk.download('averaged_perceptron_tagger_eng')
tagged_tokens = pos_tag(tokens)
print(tagged_tokens)
```

```python
print("Name: Netra Neman", "UIN: 231A013")
nltk.download('tagsets_json')
nltk.help.upenn_tagset("NNP")
```

```python
print("Name: Netra Neman", "UIN: 231A013")
pos_tag(["are"])
```

```python
print("Name: Netra Neman", "UIN: 231A013")
nltk.download('maxent_ne_chunker_tab')
named_entities = ne_chunk(tagged_tokens)
print(named_entities)
```

```python
print("Name: Netra Neman", "UIN: 231A013")

for subtree in named_entities.subtrees():
  if isinstance(subtree, nltk.Tree):
    entity = " ".join([token for token, pos in subtree.leaves()])
    label = subtree.label()
    print(f"Entity : {entity}, Label: {label}")
```
