```python
print("Name: Netra Neman", "UIN: 231A013")
morphological_info = {}
for token, lema in zip(tokens, lemas):
    if token != lema:
        if token.endswith('ing'):
            analysis = "Inflectional: Present participle/continous form of '{}'".format(lema)
        elif token.endswith("er") or token.endswith("ation"):
          analysis = "Derivational: '{}' + -er -> person who does the action".format(lema)
        else:
          analysis = "Inflectional or irregular form of '{}'".format(lema)
    else:
      analysis = "Base/root form"
    morphological_info[token]=analysis
```

```python
print("Name: Netra Neman", "UIN: 231A013")
print(morphological_info)
```
