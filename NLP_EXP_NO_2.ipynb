```python
print("Name: Netra Neman","UIN: 231A013")
import pandas as pd
def process_tweet(tweet):
  tokenizer = TweetTokenizer(preserve_case=False, strip_handles=True, reduce_len=True)

  # Convert tweet to string if it's a pandas Series
  tweet_str = tweet.item() if isinstance(tweet, pd.Series) else tweet

  tweet_cleaned = re.sub(r'^RT[\s]+', '', tweet_str)
  tweet_cleaned = re.sub(r'https?://[^\s\n\r]+', '', tweet_cleaned)
  tweet_cleaned = re.sub(r'#', '', tweet_cleaned)
  tweet_cleaned = re.sub(r'@', '', tweet_cleaned)
  tweet_cleaned = re.sub(r'\\?', '', tweet_cleaned)
  tweet_cleaned = re.sub(r'!', '', tweet_cleaned)
  tweet_cleaned = re.sub(r'[:)|:()]', '', tweet_cleaned)

  tweet_tokens = tokenizer.tokenize(tweet_cleaned)
  stopwords_english = stopwords.words('english')
  tweets_clean = []
  for word in tweet_tokens:
    if (word not in stopwords_english and word not in string.punctuation):
      tweets_clean.append(word)
  return tweets_clean
```
